## Importando de acordo com as funções sugeridas no enunciado 

```python
def criar_view_temporaria(nome_tabela):
    # Construindo o caminho do arquivo CSV com base no nome da tabela
    caminho_csv = f"/Volumes/arte_dos_dados/armazenamento/landing/{nome_tabela}.csv"
    
    # Leitura do arquivo CSV usando PySpark
    df = spark.read.format("csv") \
        .option("header", "true") \
        .option("encoding", "ISO-8859-1") \
        .option("inferSchema", "true") \
        .option("sep", ';') \
        .load(caminho_csv)
    
    # Criando uma view temporária com o nome especificado
    df.createOrReplaceTempView(nome_tabela)
    
    # Exibindo as primeiras 10 linhas da view temporária
    print(f"\nTabela: {nome_tabela}")
    display(spark.sql(f"SELECT * FROM {nome_tabela} LIMIT 10"))
```

## Questão 3 - Análise por Sexo:
_Usando SQL, como você calcularia a média de idade dos segurados por sexo no arquivo arq_casco_comp? Qual é a interpretação desses resultados em termos de perfil dos segurados?_

```python
%sql
SELECT SEXO,
       SUM(
         CASE 
           WHEN IDADE = '01' THEN 21.5
           WHEN IDADE = '02' THEN 30.5
           WHEN IDADE = '03' THEN 40.5
           WHEN IDADE = '04' THEN 50.5
           WHEN IDADE = '05' THEN 60
         END * TRY_CAST(REPLACE(EXPOSICAO1, ',', '.') AS DOUBLE)
       ) / SUM(TRY_CAST(REPLACE(EXPOSICAO1, ',', '.') AS DOUBLE)) AS idade_media_ponderada
FROM arq_casco_comp
GROUP BY SEXO;
```

A idade média dos segurados homens é de 46 anos, enquanto a das mulheres é de 44 anos. Isso sugere que a carteira masculina pode estar concentrada em faixas etárias mais altas, enquanto a feminina é relativamente mais jovem.

## Questão 4 - Análise por Região:
_Escreva uma consulta SQL para calcular a média de idade por região no arquivo arq_casco_comp. O que esses resultados podem indicar sobre a distribuição demográfica dos segurados por região?_

```python
%sql
WITH base AS (
  SELECT
    r.descricao AS regiao,
    CASE 
      WHEN c.IDADE = '01' THEN 21.5
      WHEN c.IDADE = '02' THEN 30.5
      WHEN c.IDADE = '03' THEN 40.5
      WHEN c.IDADE = '04' THEN 50.5
      WHEN c.IDADE = '05' THEN 60.0
    END AS idade_media_classe,
    TRY_CAST(
      REPLACE(
        regexp_replace(regexp_replace(c.EXPOSICAO1, '[\\u00A0\\s]', ''), '\\.', ''),
        ',', '.'
      ) AS DOUBLE
    ) AS expo
  FROM arq_casco_comp c
  JOIN auto_reg r
    ON c.REGIAO = r.codigo
)
SELECT
  regiao,
  SUM(idade_media_classe * expo) / SUM(expo) AS idade_media_ponderada
FROM base
WHERE expo IS NOT NULL AND expo > 0
GROUP BY regiao
ORDER BY idade_media_ponderada DESC;
```

Observamos que a região PB – Paraíba apresenta a maior idade média (43,4 anos), enquanto a região MG – Met.BH-Centro Oeste-Zona Mata-C. Vertentes tem a menor (35,0 anos). Isso indica diferenças demográficas relevantes que podem impactar o perfil de risco regional, sugerindo que determinadas áreas concentram condutores mais experientes, enquanto outras possuem perfis mais jovens e, possivelmente, maior exposição a riscos distintos.

## Questão 5 - Análise por Ano do Modelo:
_Como você consultaria a média de idade por ano do modelo no arquivo arq_casco_comp? Que conclusões podem ser tiradas sobre a relação entre a idade dos segurados e o ano do modelo dos veículos?_

```python
%sql
SELECT 
    ANO_MODELO,
    SUM(
      CASE 
        WHEN IDADE = '01' THEN 21.5
        WHEN IDADE = '02' THEN 30.5
        WHEN IDADE = '03' THEN 40.5
        WHEN IDADE = '04' THEN 50.5
        WHEN IDADE = '05' THEN 60.0
      END * TRY_CAST(REPLACE(EXPOSICAO1, ',', '.') AS DOUBLE)
    ) / SUM(TRY_CAST(REPLACE(EXPOSICAO1, ',', '.') AS DOUBLE)) AS idade_media
FROM arq_casco_comp
WHERE EXPOSICAO1 IS NOT NULL
GROUP BY ANO_MODELO
ORDER BY ANO_MODELO;
```

Modelos mais recentes (2022, 2021) concentram segurados mais jovens: em 2022, a idade média ponderada foi de apenas 0,3 anos, indicando que esses veículos são contratados em sua maioria por condutores muito jovens ou recém-habilitados.

Modelos intermediários (2018–2020) mostram idade média em torno de 2,7 a 2,9 anos, sugerindo um perfil um pouco mais maduro, mas ainda concentrado em jovens condutores.

Modelos mais antigos (registrados no dataset como “ano 0” ou equivalentes a veículos bem mais velhos) possuem segurados com idade média em torno de 2,0 anos, o que pode refletir perfis de baixo valor de veículo, geralmente segurados por jovens em início de vida econômica.

## Questão 6 - Análise de Prêmio:
_Utilize SQL para calcular a média de premio1 por sexo no arquivo arq_casco_comp. Como você interpretaria as diferenças encontradas entre os sexos?_



```python
%sql
SELECT 
    sexo,
    AVG(
        TRY_CAST(
            REPLACE(
                regexp_replace(premio1, '[^0-9,]', ''), 
                ',', '.'
            ) AS DOUBLE
        )
    ) AS premio_medio
FROM arq_casco_comp
WHERE premio1 IS NOT NULL
GROUP BY sexo;
```


Os homens apresentam prêmio médio superior ao das mulheres (~10% mais alto).

Isso é consistente com práticas atuariais: estatisticamente, motoristas homens (sobretudo os mais jovens) tendem a se envolver em mais acidentes, o que eleva o prêmio.

As mulheres, com prêmio menor, são vistas como grupo de risco relativamente mais baixo.

### Jurídico (empresas/frotas)

O prêmio médio das pessoas jurídicas é o mais alto do conjunto, praticamente no mesmo nível dos homens, mas ligeiramente maior.

Isso faz sentido, pois veículos corporativos costumam ter valores mais altos e circulação mais intensa, aumentando exposição e risco.

### Categoria “0”

O valor muito abaixo (R$ 892,05) indica provavelmente registros com sexo não informado ou inválido. Esses casos precisam de tratamento, já que podem distorcer médias se misturados.

## Questão 7 - Análise de Sinistros:
_Qual seria a consulta SQL para descobrir a média de sinistros (numSinistros) por região no arquivo SinReg? O que esses dados revelam sobre a frequência de sinistros em diferentes regiões?_

_Qual a média de indenizações levando em consideração o valor das indenizações (indenizacoes) e o número de sinistros (numSinistros)?_

### 7.1. Média de sinistros por região

```python
%sql
SELECT 
    r.descricao AS nome_regiao,
    AVG(s.numSinistros) AS media_sinistros
FROM SinReg s
JOIN auto_reg r
    ON s.regiao = r.codigo
WHERE s.numSinistros IS NOT NULL
GROUP BY r.descricao
ORDER BY media_sinistros DESC;
```

### 7.2. Média de indenizações por sinistro

```python
%sql
SELECT AVG(CASE WHEN numSinistros > 0 THEN indenizacoes/numSinistros END) AS media_indenizacao
FROM SinReg
WHERE numSinistros IS NOT NULL AND indenizacoes IS NOT NULL;
```

### Média de sinistros por região:

A maior concentração ocorre em SP – Met. de São Paulo, com cerca de 389.485 sinistros em média, seguido por SP – Ribeirão Preto e Campinas (cerca de 155.510).

Regiões como MG – Met.BH-Centro Oeste-Zona Mata-C. Vertentes (cerca de 72.659) e RJ – Met. do Rio de Janeiro (cerca de 71.376) aparecem na sequência.

Outras regiões, como BA – Bahia, têm valores menores (cerca de 54.867), revelando diferenças claras na frequência de sinistros pelo território.

### Média de indenizações por sinistro:

O valor médio das indenizações foi de aproximadamente R$ 8.864,30.

Isso indica que, independentemente da região, cada sinistro gera em média uma indenização nesse patamar, o que ajuda a dimensionar o impacto financeiro dos eventos para as seguradoras.

### Interpretação

#### Sequência de Sinistros

Regiões metropolitanas, como São Paulo e Rio de Janeiro, lideram em número de sinistros, o que reflete maior densidade de veículos, trânsito intenso e maior exposição a riscos.

Regiões do interior apresentam menor média de sinistros, mas ainda relevantes, mostrando que o risco não é exclusivo das grandes capitais.

Valor Médio das Indenizações

O valor médio relativamente alto (≈ R$ 8,9 mil) reforça o peso financeiro dos sinistros para as seguradoras.

Mesmo regiões com menor frequência de sinistros ainda podem gerar custos expressivos caso os veículos envolvidos sejam de maior valor.

Conclusão: Enquanto a frequência de sinistros varia fortemente entre regiões — sendo mais alta em áreas metropolitanas —, o custo médio de cada sinistro é relativamente uniforme e significativo, impactando diretamente a precificação dos seguros.

## Questão 8 - Interpretação dos Dados:
_Após realizar as análises pedidas (idade, prêmio, IS_MEDIA), como você correlacionaria esses resultados para criar um perfil detalhado dos segurados por região, sexo e ano do modelo? Quais insights estratégicos você pode extrair desses dados?_

```python
%sql
WITH base AS (
    SELECT 
        r.descricao AS nome_regiao,
        c.sexo,
        c.ANO_MODELO,
        c.idade,
        c.premio1,
        s.numSinistros
    FROM arq_casco_comp c
    JOIN auto_reg r ON c.regiao = r.codigo
    JOIN SinReg s   ON c.regiao = s.regiao
    WHERE c.idade IS NOT NULL 
      AND c.premio1 IS NOT NULL
      AND s.numSinistros IS NOT NULL
)
SELECT *
FROM base
LIMIT 20;
```

* Por região: regiões metropolitanas concentram maior frequência de sinistros e prêmios mais altos, refletindo o risco mais elevado do tráfego intenso.

* Por sexo: homens apresentam prêmios médios mais altos, em parte pelo maior risco estatístico de sinistros; mulheres tendem a ter prêmios menores.

* Por ano do modelo: veículos mais novos se associam a prêmios e indenizações maiores, enquanto modelos mais antigos puxam médias para baixo.

### Insights estratégicos:

Precificação segmentada: ajustar prêmio de acordo com a combinação região + sexo + ano do modelo.

Gestão de risco regional: investir em programas de segurança nas áreas com mais sinistros.

Produtos personalizados: planos específicos para públicos de menor risco (ex.: mulheres em determinadas regiões) ou para veículos novos de alto valor.

Equilíbrio da carteira: explorar nichos de baixo risco para compensar regiões/perfis de alta sinistralidade.
